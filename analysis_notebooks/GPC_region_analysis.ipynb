## Distributions of functional scores for different regions
Plot the distribution of averaged functional scores per site (except stop codons) stratified by region of GPC to show more mutationally tolerant and intolerant regions.

```python
# Load data and remove wildtype
functional_scores = pd.read_csv(dms_scores)
functional_scores = (
    functional_scores.loc[functional_scores["wildtype"] != functional_scores["mutant"]]
)

# Label regions as SSP, GP1, or GP2
functional_scores["region"] = (
    functional_scores.apply(
        lambda x: "SSP" if x["site"] <= 58 else ("GP1" if x["site"] <= 251 else "GP2"), axis=1
    )
)


# Add GP2 transmembrane domain 
# 420 - 442 Mann et al 2025
# 425 - 439 Taylor et al 2025
TM_sites = (
    functional_scores.loc[functional_scores["site"].isin(TM)].copy()
)
TM_sites["region"] = "TM"
functional_scores = (
    pd.concat([
        functional_scores, 
        TM_sites,
    ], ignore_index = True)
)

# Add NFP and IFL
NFP_sites = (
    functional_scores.loc[functional_scores["site"].isin(NFP)].copy()
)
NFP_sites["region"] = "NFP"
functional_scores = (
    pd.concat([
        functional_scores, 
        NFP_sites,
    ], ignore_index = True)
)
IFL_sites = (
    functional_scores.loc[functional_scores["site"].isin(IFL)].copy()
)
IFL_sites["region"] = "IFL"
functional_scores = (
    pd.concat([
        functional_scores, 
        IFL_sites,
    ], ignore_index = True)
)

# Add TfR1 binding sites
TfR1_binding_sites = (
    functional_scores.loc[functional_scores["site"].isin(TfR1_sites)].copy()
)
TfR1_binding_sites["region"] = "TfR1 binding sites"
functional_scores = (
    pd.concat([
        functional_scores, 
        TfR1_binding_sites,
    ], ignore_index = True)
)

# Add glycans
glycan_sites = (
    functional_scores.loc[functional_scores["site"].isin(glycans)].copy()
)
glycan_sites["region"] = "N-glycosylation sites"
functional_scores = (
    pd.concat([
        functional_scores, 
        glycan_sites,
    ], ignore_index = True)
)
```

```python
def plot_GPC_region_effects(stat_col, df):

    # Plot score distrbutions for each region
    distribution_plot = alt.Chart(
            df, title=f"{stat_col}"
        ).mark_circle(opacity=0.15, size=75).encode(
        y=alt.Y(
            "region:N",
            title="GPC region",
            sort=None,
            axis=alt.Axis(
                domainWidth=1,
                domainColor="black",
                tickColor="black",
            ),
        ),
        x=alt.X(
            stat_col+":Q",
            title=stat_col,
            axis=alt.Axis(
                values=[-5,-4,-3,-2,-1,0,1],
                domainWidth=1,
                domainColor="black",
                tickColor="black",
            ),
        ),
        yOffset="jitter:Q",
        color=alt.Color(
            "region:N",
            scale=alt.Scale(
                domain=functional_scores["region"].unique().tolist(), 
                range=tol_muted_adjusted
            ),
        ).legend(None),
        tooltip=[
            "site",
            "wildtype",
            "mutant",
            alt.Tooltip(
                stat_col, format=".2f", title="effect on cell entry"
            ),
        ],
    ).transform_calculate(
        # Generate Gaussian jitter with a Box-Muller transform
        jitter="sqrt(-2*log(random()))*cos(2*PI*random())"
    )
    
    # Plot median line
    median_plot = alt.Chart(df).mark_tick(size=45, thickness=3, color="#000000").encode(
        y=alt.Y(
            "region:N",
            title="GPC region",
            sort=None,
            axis=alt.Axis(
                domainWidth=1,
            ),
        ),
        x=alt.X(
            "median("+stat_col+"):Q",
            axis=alt.Axis(
                values=[-5,-4,-3,-2,-1,0,1],
                domainWidth=1,
            ),
        ),
        tooltip=[
            "region:N",
            alt.Tooltip(
                "median("+stat_col+"):Q", format=".2f", title="median effect on cell entry",
            ),
        ],
    )
    
    # Combine striplot and median
    combined_plot = (
        (distribution_plot + median_plot)
        .configure_axis(
            grid=False,
            labelFontSize=16,
            titleFontSize=16,
            labelFontWeight="normal",
            titleFontWeight="normal",
        )
        .properties(
            width=400, 
            height=400,
        )
        .configure_title(
            fontSize=24,
        ).configure_view(
            stroke=None
        )
    )
    
    # # Make output dir if doesn't exist
    # if not os.path.exists(html_dir):
    #     os.mkdir(html_dir)
    
    # print(f"Saving to {html_output}")
    # combined_plot.save(html_output)
    
    return combined_plot
```

```python
effects = [
    "entry in C musculinus TfR1 cells",
    "entry in C callosus TfR1 cells",
]
GPC_region_plots = []

for effect in effects:
    GPC_region_plots.append(plot_GPC_region_effects(effect, functional_scores))
```

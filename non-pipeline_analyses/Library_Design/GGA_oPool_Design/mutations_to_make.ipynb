```python
# Imports
import pandas as pd
```

```python
# Data

# Junin GPC codon optimized sequence 
wt_sequence = "MGQFISFMQEIPTFLQEALNIALVAVSLIAIIKGIVNLYKSGLFQFFVFLALAGRSCTEEAFKIGLHTEFQTVSFSMVGLFSNNPHDLPLLCTLNKSHLYIKGGNASFMISFDDIEVLLPQYDVIIQHPADMSWCSKSDDQIWLSQWFMNAVGHDWHLDPPFLCRNRTKTEGFIFQVNTSKTGVNENYAKKFKTGMHHLYREYPDSCLNGKLCLMKAQPTSWPLQCPLDHVNTLHFLTRGKNIQLPRRSLKAFFSWSLTDSSGKDTPGGYCLEEWMLVAAKMKCFGNTAVAKCNLNHDSEFCDMLRLFDYNKNAIKTLNDETKKQVNLMGQTINALISDNLLMKNKIRELMSVPYCNYTKFWYVNHTLSGQHSLPRCWLIKNNSYLNISDFRNDWILESDFLISEMLSKEYSDRQGKTPLTLVDICFWSTVFFTASLFLHLVGIPTHRHIRGEACPLPHRLNSLGGCRCGKYPNLKKPTVWRRRH"

# Amino acid mutations
AA_muts = ['R','K','H','D','E','Q','N','S','T','Y','W','F','A','I','L','M','V','G','P','C','*']
```

```python
# Initialize dataframe
mut_df = pd.DataFrame({
    "sequential_site" : list(range(1,len(wt_sequence)+1)), 
    "reference_site" : list(range(1,len(wt_sequence)+1)),
    "wildtype_aa" : list(wt_sequence),
    "reference_aa" : list(wt_sequence),
})

# create list of all possible mutations including stop codons
mut_df["mutant_aa"] = (
    mut_df.apply(lambda x: [mut for mut in AA_muts if mut != x["wildtype_aa"]], axis=1)
)
mut_df = mut_df.explode("mutant_aa")

# Set representation of mutations
mut_df["representation"] = (
    mut_df["mutant_aa"].apply(lambda x: 1 if x != "*" else 0)
)
# Alternate every 3rd amino acid for first 100 positions
mut_df["representation"] = (
    mut_df.apply(lambda x: 1 if ((x["sequential_site"]-1) % 3 == 0 and x["mutant_aa"] == "*" and x["sequential_site"] < 100) else x["representation"], axis=1)
)

mut_df
```

```python
# Save df
mut_df.to_csv("Data/JUNV_mutations_to_make.csv", index=False)
```
